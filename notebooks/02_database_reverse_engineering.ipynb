| STATUS | Meaning                                                                                     |
| ------ | ------------------------------------------------------------------------------------------- |
| **X**  | No loan balance available for that month (loan not yet started or no information reported). |
| **C**  | Loan closed during that month.                                                              |
| **0**  | Loan active, no days past due (paid on time).                                               |
| **1**  | 1–30 days past due.                                                                         |
| **2**  | 31–60 days past due.                                                                        |
| **3**  | 61–90 days past due.                                                                        |
| **4**  | 91–120 days past due.                                                                       |
| **5**  | More than 120 days past due (severe delinquency/default risk).                              |


```python
from src.feature_engineering.bureau_balance_features import BureauBalanceFeatureEngineer

engineer = BureauBalanceFeatureEngineer()

bb_features = engineer.transform(bureau_balance)

bb_features.head()
```

### Integrating Bureau balance and bureau together

```python
bureau_enriched = bureau.merge(
    bb_features,
    on="SK_ID_BUREAU",
    how="left"
)

bureau_enriched.head()
```
