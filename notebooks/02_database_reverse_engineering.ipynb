```python
from src.modeling.splitter import DataSplitter
from src.utils.config import config

# Start from the original merged dataset
X_raw = final_dataset.drop(columns=["TARGET"]).copy()
y = final_dataset["TARGET"].copy()

# Remove identifier columns
identifier_columns = config["preprocessing"]["identifier_columns"]
X_raw = X_raw.drop(columns=identifier_columns, errors="ignore")

# Split before imputation, encoding, and feature selection
X_train_raw, X_valid_raw, y_train, y_valid = DataSplitter.split(
    X_raw, y
)

print("Training features:", X_train_raw.shape)
print("Validation features:", X_valid_raw.shape)

print("\nTarget distribution:")
print("Training:", y_train.mean())
print("Validation:", y_valid.mean())
```

```python
import numpy as np

from src.modeling.preprocessor import ModelPreprocessor

# Reduce numerical memory usage
numeric_columns = X_train_raw.select_dtypes(
    include=["number"]
).columns

X_train_raw[numeric_columns] = (
    X_train_raw[numeric_columns].astype(np.float32)
)

X_valid_raw[numeric_columns] = (
    X_valid_raw[numeric_columns].astype(np.float32)
)

# Create preprocessor
preprocessor = ModelPreprocessor.create(X_train_raw)

# Fit ONLY on training data
X_train_processed = preprocessor.fit_transform(X_train_raw)

# Apply learned transformations to validation data
X_valid_processed = preprocessor.transform(X_valid_raw)

print("Training shape:", X_train_processed.shape)
print("Validation shape:", X_valid_processed.shape)

print("Training missing values:", np.isnan(X_train_processed).sum())
print("Validation missing values:", np.isnan(X_valid_processed).sum())
```

```python
print(
    "Training matrix memory:",
    round(X_train_processed.nbytes / (1024 ** 3), 2),
    "GB"
)
```

```python
from src.modeling.trainer import ModelTrainer
import time

model_names = [
    "logistic_regression",
    "random_forest",
    "xgboost",
    "lightgbm",
    "catboost"
]

trained_models = {}

for name in model_names:

    print(f"\nTraining {name}...")
    start_time = time.time()

    model = ModelTrainer.train(
        name,
        X_train_processed,
        y_train
    )

    trained_models[name] = model

    elapsed = round(time.time() - start_time, 2)

    print(f"{name} trained successfully in {elapsed} seconds.")
```

```python
import pandas as pd

from src.modeling.evaluator import ModelEvaluator

evaluation_results = {}
confusion_matrices = {}

for name, model in trained_models.items():

    print(f"Evaluating {name}...")

    evaluation = ModelEvaluator.evaluate(
        model,
        X_valid_processed,
        y_valid,
        threshold=0.5
    )

    evaluation_results[name] = evaluation["metrics"]
    confusion_matrices[name] = evaluation["confusion_matrix"]

# Create comparison table
results_df = pd.DataFrame(evaluation_results).T

results_df = results_df.sort_values(
    by="PR_AUC",
    ascending=False
)

results_df.round(4)
```
